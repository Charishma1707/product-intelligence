# Dataset Review (Compact & Clean View)
Adjusted to normal print size with vertical field inspection for easy readability.

```python
import os
import pandas as pd
import IPython.display as display
from IPython.display import HTML

# Reset to normal, clean, compact display sizes
pd.set_option('display.max_columns', 20)
pd.set_option('display.max_rows', 50)
pd.set_option('display.max_colwidth', 60)
pd.set_option('display.width', 1000)

# Helper function to find CSV path regardless of working directory
def find_csv(filename):
    candidates = [
        filename,
        os.path.join("..", filename),
        os.path.join("product-intelligence", filename),
        os.path.join("..", "product-intelligence", filename),
        r"C:\charishma\apicurio registry\unihack\product-intelligence\" + filename
    ]
    for p in candidates:
        if os.path.exists(p):
            return p
    return filename

expected_path = find_csv("Unihack_ Expected Output - Delivery Format.csv")
input_path = find_csv("Unihack_ Sample Dataset - Input.csv")

print("Paths resolved successfully.")
```

### 1. Expected Output Delivery Format (Normal Print Size + Vertical Field Inspector)

```python
df_expected = pd.read_csv(expected_path, encoding='utf-8-sig', engine='python')
print(f"Expected Output Dataset: {df_expected.shape[0]} rows x {df_expected.shape[1]} columns\n")

# Show key core columns in standard compact view
key_cols = [c for c in ['Part_Manuf', 'Mfg_Part_Num', 'Part_Desc', 'MANUFACTURER_NAME', 'Classpath', 'SHORT_DESC', 'UNSPSC', 'Product Image'] if c in df_expected.columns]
print("--- Key Identity Columns ---")
display.display(df_expected[key_cols].head())

print("\n--- Row 1 All Fields (Vertical View for Easy Reading) ---")
# Transpose row 0 so you can read all 252 fields vertically instead of endless horizontal scrolling
row1_vertical = df_expected.iloc[[0]].T.dropna()
row1_vertical.columns = ['Sample Expected Value']
display.display(row1_vertical.head(35))
```
